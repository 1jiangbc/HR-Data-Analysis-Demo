```python
#series题目1
import numpy as np
import pandas as pd
np.random.seed(42)
scores  = pd.Series(np.random.randint(50,101,10),index=['学生'+str(i)for i in range(1,11)])
print(scores)
print(scores.values)
print('平均分',scores.values.mean())
print('最高分',scores.values.max())
print('最低分',scores.values.min())
#高于平均分的人数
mean = scores.mean()

print('高于平均分学生的人数',scores[scores>mean].count())
```

```python
#series题目2
temperature = pd.Series([28,31,29,32,30,27,33],index=['周一','周二','周三','周四','周五','周六','周日'])
print(temperature[temperature>30])
print('平均温度',temperature.mean())
#如果我只要温度丢掉索引呢？
#print(sorted(temperature.values,reverse=True))
print(temperature.sort_values(ascending=False))#ascending就是“升序”（从小到大）的意思。

# 用 idxmax() 拿到最高温那天的索引，idxmin() 拿最低温那天的索引
print(f'温度最高的一天{temperature.idxmax()}，温度是{temperature.max()}')
print(f'温度最低的一天{temperature.idxmin()}，温度是{temperature.min()}')

#找出温度变化最大的两天
t3 = temperature.diff().abs()
t3.sort_values(ascending=False).index#等于key
print('温度变化量最大的两天',t3.sort_values(ascending=False).index[:2].tolist())
```

```python
#series题目3
prices = pd.Series([102.3,103.5,105.1,104.8,106.2,107.0,106.5,108.1,109.3,110.2],index = pd.date_range('2023-1-01',periods=10))
arr = []
for i in range(len(prices)-1):
    arr.append(prices.iloc[i+1]/prices.iloc[i]-1)
#print(arr)
result = pd.Series(arr,index=prices.index[1:])
#print(result)
print('收益率最高的日期',result.idxmax())
print('收益率最低的日期',result.idxmin())
print('波动率',result.std())
```

```python
#series题目4
sales = pd.Series([120,135,145,160,155,170,180,175,190,200,210,220],index=pd.date_range('2022-01-01',periods=12,freq='ME'))
arr = []
for i in range(len(sales)//3):
    arr.append((sales.iloc[i*3:i*3+3]).mean())
print(arr)

print(sales.idxmax())

month_money = []
for i in range(len(sales)-1):
    month_money.append((sales.iloc[i+1]-sales.iloc[i])/sales.iloc[i])

print('月环比增长率',month_money)
month_money = pd.Series(month_money,sales.index[1:])

up_month = []  # 用你截图里写的变量名
count = 0      # 状态计数器

for i in range(len(month_money) - 1):
    curr = month_money.iloc[i]
    nxt = month_money.iloc[i+1]

    if curr > 0 and nxt > 0:
        if count == 0:
            # 新窗口开启，双存
            up_month.append(month_money.index[i])
            up_month.append(month_money.index[i+1])
            count = 2
        else:
            # 窗口延续，只存右边
            up_month.append(month_money.index[i+1])
            count += 1
    else:
        # 窗口关闭，计数器归零
        count = 0

print("连续增长的月份有：")
for m in up_month:
    print(m)
```

```python
##dataframe的创建方式
import pandas as pd
import numpy as np
#通过series创建
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([6,7,8,9,10])
df = pd.DataFrame({'第一列':s1,'第二列':s2})
#print(df)
#通过字典创建
df = pd.DataFrame(
    {
        "id":[1,2,3,4,5],
        "name":["tom","jack","alice","bob","allen"],
        "age":[15,17,20,26,30]
    },index=[1,2,3,4,5],columns=["id","name","age"]
)



print(df)
```

```python
#dataframe常用方法
print(df.head())#查看前n行默认5行
print(df.tail(1))#查看后1行
#print(df.isin(['jack']))#判断元素在不在数组里，返回布尔类型
print(df.isin(['jack',20]))#确切地说就是：扫描整个表格里的每一个格子，判断每个格子的值，是否等于 'jack' 或者 20。如果是，那个格子就标记为 True；不是，就标记为 False。

print(df.isna())#查看是否有缺失值

print(df['age'].sum())
print(df.age.max())
print(df.age.min())
print(df['age'].mean())
print(df['age'].mode())#都是唯一值就返回一个表


print(df['age'].std())
print(df['age'].var())
print(df['age'].quantile(0.25))#排在25%位置的数

print(df.describe())


print(df.count())#每一列非缺失值的个数

print(df.value_counts())#出现的次数，查看有没有重复行

print(df.drop_duplicates())#去重

print(df.duplicated())#返回的是一个布尔dataframe，但是不去重复，不是重复项标记为true

print(df.duplicated(subset=['age']))#查看是否重复，subject指的是要坚持列，age是列名

df.sample(2)#随机抽样

df.cumsum()
df.iloc[:,2].cummax()#累计最大

print(df.sort_index(ascending=False))

print(df.sort_values(by=['age'],ascending=False))

df.nlargest(2,columns=['age'])
```

```python
#案列分析
data = {
    '姓名':['张三','李四','王五','赵六','钱七'],
    '数学':[85,92,78,88,95],
    '英语':[90,88,85,92,80],
    '物理':[75,80,88,85,90]
}

data = pd.DataFrame(data)
#总分和平均分
print('总分',data.iloc[:,1:4].sum(axis=1))
print('平均分',data.iloc[:,1:4].mean(axis=1))

#找出数学成绩高于90或英语成绩高于85的人
print(data[(data['数学']>90)|(data['英语']>85)])


#总分从高到低，取前三个

# 1. 将总分算出来，并真正“写入”到data表格的新列中
data['总分'] = data.iloc[:, 1:4].sum(axis=1)

# 2. 现在表里有“总分”这一列了，开始排序取前三
top3 = data.sort_values(by='总分', ascending=False).head(3)
print(top3)
```
