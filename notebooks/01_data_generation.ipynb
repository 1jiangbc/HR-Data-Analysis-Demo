```python
import pandas as pd
import numpy as np
import os

# 设置随机种子，保证每次生成的数据一样
np.random.seed(42)
n = 400

# 1. 构建基础数据
data = {
    'person_id': np.arange(1, n + 1),
    'gender': np.random.choice(['Male', 'Female'], n),
    'age': np.random.randint(18, 71, n),
    'occupation': np.random.choice(['Manual Labor', 'Retired', 'Office Worker', 'Student', 'Sales'], n),
    'sleep_duration': np.round(np.random.uniform(4.0, 11.0, n), 1),
    'sleep_quality': np.round(np.random.uniform(3.0, 10.0, n), 1),
    'physical_activity_level': np.random.randint(20, 101, n),
    'stress_level': np.random.randint(1, 11, n),
    'bmi_category': np.random.choice(['Normal', 'Obese', 'Overweight', 'Underweight'], n),
    'blood_pressure': [f"{np.random.randint(110, 140)}/{np.random.randint(70, 95)}" for _ in range(n)],
    'heart_rate': np.random.randint(60, 101, n),
    'daily_steps': np.random.randint(1000, 15000, n)
}

# 2. 特殊处理：sleep_disorder 只有 110 个非空值
# 先生成 290 个 NaN，然后拼上 110 个具体的疾病标签
disorders = np.random.choice(['None', 'Insomnia', 'Sleep Apnea'], 110)
sleep_disorder = [np.nan] * 290 + list(disorders)
np.random.shuffle(sleep_disorder)  # 打乱顺序
data['sleep_disorder'] = sleep_disorder

# 3. 组装 DataFrame
df = pd.DataFrame(data)

# 4. 确保文件夹存在并保存
os.makedirs('data', exist_ok=True)
df.to_csv('data/sleep.csv', index=False, encoding='utf-8-sig')

print("✅ 数据生成完毕！共", len(df), "行。")
```
